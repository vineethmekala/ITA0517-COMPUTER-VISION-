```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((400, 600, 3), dtype=np.uint8) * 255

# Draw rectangle
cv2.rectangle(img, (150, 100), (450, 300), (255, 0, 0), 3)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Rectangle Shape")
plt.axis('off')
plt.show()
```
