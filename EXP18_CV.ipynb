```python

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Read image
filename = next(iter(uploaded))
img = cv2.imread(filename)

if img is None:
    print("Error: Unable to read image")
else:
    # Get image dimensions
    height, width = img.shape[:2]

    # Select ROI based on image size
    x = width // 4
    y = height // 4
    w = width // 2
    h = height // 2

    # Crop image
    roi = img[y:y+h, x:x+w]

    # Copy ROI
    copied_roi = roi.copy()

    # Paste ROI into another location
    output = img.copy()

    x2 = 0
    y2 = 0

    output[y2:y2+h, x2:x2+w] = copied_roi

    # Display results
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
    plt.title("Cropped ROI")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
    plt.title("Copied and Pasted Image")
    plt.axis("off")

    plt.show()

    # Save output
    cv2.imwrite("roi_output.jpg", output)

    # Download image
    files.download("roi_output.jpg")
```
