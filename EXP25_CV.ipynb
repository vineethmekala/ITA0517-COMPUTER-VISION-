```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload original image
print("Upload Original Image")
uploaded = files.upload()
image_name = list(uploaded.keys())[0]

# Upload watch reference image
print("Upload Watch Reference Image")
uploaded = files.upload()
template_name = list(uploaded.keys())[0]

# Read images
img = cv2.imread(image_name)
template = cv2.imread(template_name)

# Convert images to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
temp = cv2.cvtColor(template, cv2.COLOR_BGR2GRAY)

# Get template dimensions
h, w = temp.shape

# Apply template matching
result = cv2.matchTemplate(
    gray, temp, cv2.TM_CCOEFF_NORMED
)

# Find best match
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

# Recognize watch
if max_val > 0.6:
    x, y = max_loc

    cv2.rectangle(
        img, (x,y), (x+w,y+h), (0,255,0), 3
    )

    print("Watch Recognized Successfully!")
else:
    print("Watch Not Found!")

# Display output
print("Output Image")
cv2_imshow(img)
```
