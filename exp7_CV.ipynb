```python
import cv2
from google.colab import files
from IPython.display import display, Video

# Upload video
uploaded = files.upload()
filename = next(iter(uploaded))

# Read video
cap = cv2.VideoCapture(filename)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print("Original FPS:", fps)
print("Video Width:", width)
print("Video Height:", height)

cap.release()

# Slow motion video (0.5x speed)
slow_video = cv2.VideoCapture(filename)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out_slow = cv2.VideoWriter(
    'slow_motion.mp4', fourcc, fps / 2, (width, height)
)

while True:
    ret, frame = slow_video.read()
    if not ret:
        break
    out_slow.write(frame)

slow_video.release()
out_slow.release()

# Fast motion video (2x speed)
fast_video = cv2.VideoCapture(filename)
out_fast = cv2.VideoWriter(
    'fast_motion.mp4', fourcc, fps * 2, (width, height)
)

while True:
    ret, frame = fast_video.read()
    if not ret:
        break
    out_fast.write(frame)

fast_video.release()
out_fast.release()

# Display videos
print("Slow Motion Video (0.5x)")
display(Video('slow_motion.mp4', embed=True))

print("Fast Motion Video (2x)")
display(Video('fast_motion.mp4', embed=True))

# Download processed videos
files.download('slow_motion.mp4')
files.download('fast_motion.mp4')
```
