```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Get image dimensions
rows, cols = img.shape[:2]

# Define three points in the original image
pts1 = np.float32([[50, 50], [200, 50], [50, 200]])

# Define three corresponding points in the transformed image
pts2 = np.float32([[50, 100], [200, 100], [100, 200]])

# Calculate affine transformation matrix
matrix = cv2.getAffineTransform(pts1, pts2)

# Apply affine transformation
result = cv2.warpAffine(img, matrix, (cols, rows))

# Display images
print("Original Image:")
cv2_imshow(img)

print("Affine Transformed Image:")
cv2_imshow(result)
```
