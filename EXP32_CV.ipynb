```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Get image size from user
width = int(input("Enter image width: "))
height = int(input("Enter image height: "))

# Create white image
img = np.ones((height, width, 3), dtype=np.uint8) * 255

# Box size
box = 100

# Draw four colored boxes
cv2.rectangle(img, (0, 0), (box, box), (0, 0, 0), -1)
cv2.rectangle(img, (width-box, 0), (width-1, box), (255, 0, 0), -1)
cv2.rectangle(img, (0, height-box), (box, height-1), (0, 255, 0), -1)
cv2.rectangle(img, (width-box, height-box),
              (width-1, height-1), (0, 0, 255), -1)

# Display image
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Four Colored Boxes")
plt.axis('off')
plt.show()
```
