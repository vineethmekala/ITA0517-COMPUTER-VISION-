```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((400, 600, 3), dtype=np.uint8) * 255

# Draw circle
cv2.circle(img, (300, 200), 100, (0, 0, 255), 3)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Circle Shape")
plt.axis('off')
plt.show()
```
